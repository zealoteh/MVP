# Projeto MVP: Pipeline de Dados - Steam
**Objetivo:** Construir uma Pipeline de dados utilizando a arquitetura Medallion para processar, limpar e modelar dados da plataforma Steam.

## 1. Extração de Dados
Realizei a ingestão dos dados brutos em formato CSV diretamente do Unity Catalog. Os dados são lidos pelo PySpark definindo os tipos de dados e já separa os dados em colunas.

```python
from pyspark.sql.functions import *

caminho_arquivo = "/Volumes/workspace/mvp/mvp/games-features.csv"

dado_bruto = (spark.read
                  .format("csv")
                  .option("header", "true") # Cabeçalho
                  .option("inferSchema", "true") #Tipos dos dados
                  .option("escape", '"') # Ignora a , se estiver entre string
                  .load(caminho_arquivo))

display(dado_bruto)
```

## 2. Qualidade e Transformação
Após a ingestão, apliquei as seguintes limpeza de dados
* Remoção de registros nulos nas chaves principais (ID e Nome Jogo).
* Exclusão de registros duplicados na coluna ID do jogo.
* Padronização da nomenclatura das colunas para o português.

```python
num_linhas = dado_bruto.count()
print(f"Total de jogos no arquivo bruto: {num_linhas}")

# Limpeza e Padronização
dado_limpo = (dado_bruto
              .filter(col("QueryID").isNotNull() & col("QueryName").isNotNull())
              .dropDuplicates(["QueryID"])
              .withColumnRenamed("QueryID", "id_jogo")
              .withColumnRenamed("QueryName", "nome_jogo")
              .withColumnRenamed("ReleaseDate", "data_lancamento")
              .withColumnRenamed("PriceFinal", "preco_final")
             )

linhas_limpas = dado_limpo.count()
print(f"Total de jogos após a limpeza: {linhas_limpas}")

display(dado_limpo)
```

## 3. Modelagem Dimensional
Para otimizar as consultas analíticas e a criação de Dashboards, separei os dados limpos em 2 tabelas dimensão e 1 tabela fato:
* **Tabela Dimensão (`dim_jogo`):** Dados descritivos dos jogos.
* **Tabela Dimensão (`dim_genero`):** Utilizei o *Unpivot* para converter as múltiplas colunas booleanas de categorias em linhas e criar uma análise da quantidade de jogos por gênero.
* **Tabela Fato (`fato_estatistica`):** Centraliza as métricas de desempenho e valores numéricos (Preço, Notas do Metacritic, Volume de Recomendações e Estimativa de Jogadores).

```python
# Extrai o ano da coluna data_lancamento
ano = regexp_extract(col("data_lancamento"), r"(\d{4})", 1)

# Criação da Tabela Dimensão com id e nome dos jogos 
dim_jogo = dado_limpo.select(
    "id_jogo",
    "nome_jogo",
    when(ano == "", None).otherwise(ano).cast("int").alias("ano_lancamento"),
    col("IsFree").alias("eh_gratuito")
)

#Transforma as colunas booleans em linhas
expressao_unpivot = """stack(7, 
    'Indie', GenreIsIndie, 
    'Ação', GenreIsAction, 
    'Aventura', GenreIsAdventure, 
    'Casual', GenreIsCasual, 
    'Estratégia', GenreIsStrategy, 
    'RPG', GenreIsRPG, 
    'Simulação', GenreIsSimulation
) as (genero, possui_genero)"""

# Criação da tabela Dimensão com os genêros em linhas
dim_genero = (dado_limpo
              .select("id_jogo", expr(expressao_unpivot))
              .filter(col("possui_genero") == True)
              .select("id_jogo", "genero")
             )

# Criação da Tabela Fato com as metricas e estatisticas dos jogos
fato_estatistica = dado_limpo.select(
    "id_jogo",
    "preco_final",
    col("Metacritic").alias("nota_metacritic"),
    col("RecommendationCount").alias("total_recomendacoes"),
    col("SteamSpyOwners").alias("total_jogadores_estimado")
)

print("--- Tabela Dimensão dos Jogos ---")
display(dim_jogo)

print("--- Tabela Dimensão de Gêneros ---")
display(dim_genero)

print("--- Tabela Fato Metricas---")
display(fato_estatistica)
```

## 4. Carga de Dados 
Salva os Dataframes processados fisicamente no banco de dados do Unity Catalog. Isso finaliza o processo de ETL e disponibiliza os dados para a camada de visualização.

```python
# Salvando a Tabela Dimensão
dim_jogo.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.mvp.dim_jogo")

# Salvando a Tabela Fato
fato_estatistica.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.mvp.fato_estatistica")

# Salvando a Tabela Dimensão de Gêneros
dim_genero.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.mvp.dim_genero")

print("Pipeline finalizado! Tabelas salvas com sucesso no Unity Catalog.")
```

## 5. Análise de Dados e Respostas de Negócio
Com os dados modelados no Unity Catalog, esta etapa valida a integridade das relações e extrai os *insights* para responder às perguntas de negócio definidas no objetivo do projeto.

### 5.1. Gráfico da Análise de Monetização
**Pergunta:** Jogos gratuitos atraem uma base de jogadores significativamente maior? Como a crítica avalia a qualidade desses jogos em comparação aos pagos?

**Validação técnica:** Esta consulta testa o relacionamento (1:1) entre a `dim_jogo` e a `fato_estatistica`.

```python
%sql
SELECT 
    CASE 
        WHEN d.eh_gratuito = True THEN 'Gratuito' 
        ELSE 'Pago' 
    END AS `Modelo de Monetização`,
    COUNT(d.id_jogo) AS total_jogos,
    ROUND(AVG(f.total_jogadores_estimado), 0) AS media_jogadores,
    ROUND(AVG(f.nota_metacritic), 1) AS media_metacritic
FROM workspace.mvp.fato_estatistica f
INNER JOIN workspace.mvp.dim_jogo d USING (id_jogo)
WHERE f.nota_metacritic > 0 
GROUP BY 1 
ORDER BY media_jogadores DESC;
```

### 5.2. Gráfico da Evolução dos preços
**Pergunta:** Como o mercado de jogos de PC se comportou financeiramente ao longo dos anos? Existe uma tendência de aumento no valor final?

**Validação técnica:** Valida o tratamento de qualidade de dados feito na extração do ano. Foram isolados apenas os jogos pagos em um recorte histórico confiável (1997-2016).

```python
%sql
SELECT 
    d.ano_lancamento,
    COUNT(d.id_jogo) AS total_lancamentos,
    ROUND(AVG(f.preco_final), 2) AS preco_medio
FROM workspace.mvp.fato_estatistica f
INNER JOIN workspace.mvp.dim_jogo d USING (id_jogo)
WHERE d.ano_lancamento BETWEEN 1997 AND 2016 
  AND d.eh_gratuito = False 
GROUP BY d.ano_lancamento
ORDER BY d.ano_lancamento;
```

### 5.3. Engajamento e Avaliação por Gênero
**Pergunta:** Quais gêneros geram a comunidade mais engajada em volume de recomendações? 

**Validação técnica:** Esta consulta atesta o funcionamento da tabela de relacionamento `dim_genero`.

```python
%sql
SELECT 
    g.genero,
    COUNT(d.id_jogo) AS total_jogos,
    ROUND(AVG(f.nota_metacritic), 1) AS media_metacritic,
    ROUND(AVG(f.total_recomendacoes), 0) AS media_recomendacoes
FROM workspace.mvp.fato_estatistica f
INNER JOIN workspace.mvp.dim_jogo d USING (id_jogo)
INNER JOIN workspace.mvp.dim_genero g USING (id_jogo)
WHERE f.nota_metacritic > 0
GROUP BY g.genero
ORDER BY media_recomendacoes DESC;
```

## 6. Conclusão e Resultados Analíticos
O Pipeline de dados foi concluído com sucesso e com isso foi possível realizar as seguintes analises:

* **Analise da Monetização:** Jogos gratuitos (*Free-to-Play*) alcançam uma base de jogadores maior (na casa dos milhões), mas mantêm um empate técnico na qualidade percebida pela crítica em comparação aos jogos pagos.
* **Inflação e Precificação:** A análise histórica revelou uma tendência de alta no mercado, saindo do período de baixa nos preços no início dos anos 2000 para atingir o pico de valorização média por volta de 2013, marcando uma nova era de precificação na indústria.
* **Domínio de Engajamento:** Embora existam muitos títulos nos nichos *Indie* e *Casual*, as comunidades mais engajadas pertencem aos gêneros tradicionais de **Ação** e **RPG**.
